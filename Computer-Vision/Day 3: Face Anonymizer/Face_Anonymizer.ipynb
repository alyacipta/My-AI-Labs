# Detect Faces

```python
import cv2
import matplotlib.pyplot as plt

# Definisikan kembali path gambar
img_path = './content/face.jpg'

# Muat gambar jika belum ada di memori
try:
    img
except NameError:
    img = cv2.imread(img_path)

# Memastikan gambar sudah terbaca dengan benar
if img is None:
    print("Gambar gagal dimuat! Periksa kembali path gambar Anda di:", img_path)
else:
    # Konversi gambar ke grayscale (diperlukan untuk deteksi wajah Haar Cascade)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Memuat classifier wajah bawaan OpenCV
    face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

    # Deteksi wajah
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

    # Buat salinan gambar untuk digambar
    annotated_image = img.copy()

    if len(faces) > 0:
        print(f"Berhasil mendeteksi {len(faces)} wajah!")
        # Gambar kotak di sekitar setiap wajah yang terdeteksi
        for (x, y, w, h) in faces:
            cv2.rectangle(annotated_image, (x, y), (x + w, y + h), (0, 255, 0), 3)

        # Tampilkan hasilnya menggunakan matplotlib
        plt.figure(figsize=(8, 6))
        plt.imshow(cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB))
        plt.axis('off')
        plt.show()
    else:
        print("Tidak ada wajah yang terdeteksi. Silakan periksa apakah file gambar valid dan memiliki wajah terfokus.")
```
