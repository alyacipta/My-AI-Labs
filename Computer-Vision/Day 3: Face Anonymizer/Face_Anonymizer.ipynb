# Read Image

```python
import urllib.request
import cv2
import numpy as np

# URL Raw dari GitHub untuk gambar Anda
img_url = 'https://raw.githubusercontent.com/alyacipta/My-AI-Labs/95b985976ce1301c5ca5043938a4a2849520b20c/Computer-Vision/Day%203%3A%20Face%20Anonymizer/images.jpg'

try:
    # Unduh gambar dari URL
    req = urllib.request.urlopen(img_url)
    arr = np.asarray(bytearray(req.read()), dtype=np.uint8)
    img = cv2.imdecode(arr, -1)

    if img is not None:
        print("Gambar berhasil diunduh dan dimuat langsung dari GitHub!")
    else:
        print("Gagal mendekode gambar dari URL.")
except Exception as e:
    print(f"Terjadi kesalahan saat mengunduh gambar: {e}")
```

# Detect Faces

```python
import sys
# Menghapus cache cv2 jika ada duplikasi atau modul rusak di memori
if 'cv2' in sys.modules:
    del sys.modules['cv2']

import cv2
import matplotlib.pyplot as plt

# Memastikan gambar sudah terbaca dengan benar dari langkah sebelumnya
if 'img' not in locals() or img is None:
    print("Gambar belum dimuat! Jalankan kembali sel di atas untuk mengunduh gambar dari GitHub.")
else:
    # Konversi gambar ke grayscale (diperlukan untuk deteksi wajah Haar Cascade)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Memuat classifier wajah bawaan OpenCV
    try:
        face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
    except AttributeError as e:
        print("Masih terjadi error pada modul cv2:", e)
        print("Solusi: Silakan klik menu 'Runtime' -> 'Restart session' di Google Colab, lalu jalankan kembali sel pengunduh gambar dan sel ini.")
        face_cascade = None

    if face_cascade is not None:
        # Deteksi wajah
        faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

        # Buat salinan gambar untuk digambar
        annotated_image = img.copy()

        if len(faces) > 0:
            print(f"Berhasil mendeteksi {len(faces)} wajah!")
            # Gambar kotak di sekitar setiap wajah yang terdeteksi
            for (x, y, w, h) in faces:
                cv2.rectangle(annotated_image, (x, y), (x + w, y + h), (0, 255, 0), 3)

            # Tampilkan hasilnya menggunakan matplotlib
            plt.figure(figsize=(8, 6))
            plt.imshow(cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB))
            plt.axis('off')
            plt.show()
        else:
            print("Tidak ada wajah yang terdeteksi. Silakan periksa apakah file gambar memiliki wajah yang terfokus.")
```
